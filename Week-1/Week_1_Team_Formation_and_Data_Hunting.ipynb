## Dataset Selection

We selected four datasets for the Moneyball: IPL Player Valuation project.

| Dataset | Format | Purpose |
|---|---|---|
| Auction | Excel (.xlsx) | Analyse player auction prices |
| Top 100 Batsmen | Excel (.xlsx) | Study batting performance |
| Top 100 Bowlers | Excel (.xlsx) | Study bowling performance |

### Why These Datasets?

- Auction data provides player price information.
- Batting data helps study runs and other batting statistics.
- Bowling data helps study wickets and bowling statistics.
- Match data provides information about match outcomes and player performances.

Combining these datasets can help us investigate which performance statistics are associated with IPL player valuation.
