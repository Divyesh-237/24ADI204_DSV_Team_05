# Week 2 — Data Audit & First Impression

## Objective
Week 2 focuses on loading the selected datasets, checking their structure and data types, identifying missing values/duplicates, and preparing the **First Impression Report**.

```python
import pandas as pd
import numpy as np

auction = pd.read_excel('../Week-1/Auction.xlsx')
matches = pd.read_csv('../Week-1/matches.csv')
batsmen = pd.read_excel('../Week-1/Top_100_batsman.xlsx')
bowlers = pd.read_excel('../Week-1/Top_100_bowlers.xlsx')

datasets = {
    'Auction': auction,
    'Matches': matches,
    'Batsmen': batsmen,
    'Bowlers': bowlers
}
```

```python
for name, df in datasets.items():
    print(f'\n{name}: {df.shape[0]} rows × {df.shape[1]} columns')
    print('Columns:', list(df.columns))
    print('Duplicate rows:', df.duplicated().sum())
    display(df.head())
```

```python
for name, df in datasets.items():
    print(f'\n--- {name}: data types ---')
    display(df.dtypes)
```

```python
print('Missing values by dataset:')
for name, df in datasets.items():
    print(f'\n{name}')
    display(df.isna().sum()[df.isna().sum() > 0].sort_values(ascending=False))
```

## First Impression Report

**Auction:** 56 rows and 5 columns. No missing values and no duplicate rows. The dataset contains IPL auction records from 2014–2020.

**Matches:** 756 rows and 18 columns. No duplicate rows. Missing values are concentrated in `umpire3`, with smaller numbers in `city`, `winner`, `player_of_match`, `umpire1`, and `umpire2`.

**Top 100 Batsmen:** 100 rows and 14 columns. No missing values and no duplicate rows. `HS` is stored as an object/text field because some scores contain `*`.

**Top 100 Bowlers:** 100 rows and 13 columns. No missing values and no duplicate rows. `BBI` should be reviewed because it has no useful variation in the supplied file.

**Initial conclusion:** The datasets are structurally usable, but the matches data requires missing-value handling and some fields require datatype/format standardisation before EDA and player-valuation modelling.
