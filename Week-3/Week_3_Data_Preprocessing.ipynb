## Cleaning Strategy

- `umpire3`: 637 missing values. Treat as not recorded; use `Unknown` if retained, or exclude it from player-valuation modelling if irrelevant.
- `city`: 7 missing values. Use `Unknown` if city analysis is required.
- `winner` and `player_of_match`: 4 missing values each. Do not blindly impute.
- `umpire1` and `umpire2`: 2 missing values each. Use `Unknown` if retained.
- Convert `date` to datetime.
- Convert batsmen `HS` to a numeric analysis field after removing `*`.
- Standardise player names before merging datasets.
- Review IPL team-name changes across seasons before joining datasets.
- Flag outliers rather than automatically deleting them because exceptional IPL performances can be genuine.
